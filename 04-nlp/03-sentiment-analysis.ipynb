1. Rule Based Sentiment Analysis
   

Not a ML method
   Polarity Score- A neumerical value to indicate if a word is positive, negative or neutral.
   It's from -1 to 1
   TextBlob and VADER both have their own different lexicon

TextBlob

```python
from textblob import TextBlob
```

```python
sentence_1 = "i had a great time at the movie it was really funny"
sentence_2 = "i had a great time at the movie but the parking was terrible"
sentence_3 = "i had a great time at the movie but the parking wasn't great"
sentence_4 = "i went to see a movie"
sentence_1_score = TextBlob(sentence_1)
sentence_2_score = TextBlob(sentence_2)
sentence_3_score = TextBlob(sentence_3)
sentence_4_score = TextBlob(sentence_4)
```

```python
print(sentence_1,sentence_1_score.sentiment.polarity)
```

```python
print(sentence_2,sentence_2_score.sentiment.polarity)
```

```python
print(sentence_3,sentence_3_score.sentiment.polarity)
```

```python
print(sentence_4,sentence_4_score.sentiment.polarity)
```

VADER

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
vader_sentiment = SentimentIntensityAnalyzer()
```

```python
print(sentence_1,vader_sentiment.polarity_scores(sentence_1))
```

```python
print(sentence_2,vader_sentiment.polarity_scores(sentence_2))
```

```python
print(sentence_3,vader_sentiment.polarity_scores(sentence_3))
```

```python
print(sentence_4,vader_sentiment.polarity_scores(sentence_4))
```

2. Pre-trained Transformer Model

```python
import transformers
from transformers import pipeline
```

```python
sentiment_pipeline = pipeline("sentiment-analysis")
```

```python
print(sentence_1)
sentiment_pipeline(sentence_1)
```

```python
print(sentence_2)
sentiment_pipeline(sentence_2)
```

```python
print(sentence_3)
sentiment_pipeline(sentence_3)
```

```python
print(sentence_4)
sentiment_pipeline(sentence_4)
```

The default model isn't giving us great results for our neutral sentence. However there are loads of models we can choose to use that have been trained on different data with different parameters: https://huggingface.co/models?pipeline_tag=text-classification&sort=downloads&search=sentiment

Let's try a specific model trained on tweets in the English language only and see if that improves our results.

```python
specific_model = pipeline("sentiment-analysis", model="finiteautomata/bertweet-base-sentiment-analysis")
```

```python
print(sentence_1)
specific_model(sentence_1)
```

```python
print(sentence_2)
specific_model(sentence_2)
```

```python
print(sentence_3)
specific_model(sentence_3)
```

```python
print(sentence_4)
specific_model(sentence_4)
```
