```python
import pandas as pd
import matplotlib.pyplot as plt
import spacy 
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import re
```

```python
bbc_news = pd.read_csv("02 bbc-news.csv")
bbc_news.head()
```

```python
bbc_news.info()
```

```python
titles = pd.DataFrame(bbc_news["title"])
titles.head()
```

```python
#lowercase
titles["lower"] = titles["title"].apply(lambda x: x.lower())
titles.head()
```

```python
en_stopword = stopwords.words("english")
en_stopword
```

```python
#stopwords removal
titles["no_stopword"] = titles["lower"].apply(lambda x : " ".join(word for word in x.split() if word not in en_stopword))
titles.head()
```

```python
#removal
pattern = r"[^\w\s]"
titles["removal"] = titles["no_stopword"].apply(lambda x : re.sub(pattern,"",x))
titles.head()
```

```python
#tokenize
titles["tokenize"] = titles["removal"].apply(lambda x : word_tokenize(x))
titles["tokenize_raw"] = titles["title"].apply(lambda x : word_tokenize(x))
titles.head()
```

```python
#lemmatize
lemmatizer= WordNetLemmatizer()
titles["lemmatize"] = titles["tokenize"].apply(lambda x : [lemmatizer.lemmatize(word) for word in x])
titles.head()
```

```python
#create list
raw_token_list = sum( titles["tokenize_raw"],[])
lemmatized_token_list = sum(titles["lemmatize"],[])
```

Parts of speech tagging

```python
nlp = spacy.load("en_core_web_sm")
```

```python
spacy_doc = nlp(" ".join(raw_token_list))
```

```python
pos = pd.DataFrame(columns= ["token", "pos_tag"])
```

```python
for word in spacy_doc:
    pos = pd.concat([pos, pd.DataFrame.from_records([{"token": word.text, "pos_tag": word.pos_ }])],ignore_index= True)
pos.head()
```

```python
#tokenfrequency
pos_token_freq = pos.groupby(["token", "pos_tag"]).size().reset_index(name = "count").sort_values(by = "count", ascending = False)
pos_token_freq.head(5)
```

```python
#most common noun
nouns =pos_token_freq[pos_token_freq.pos_tag == "NOUN"]
nouns[:10]
```

```python
#most common verb
verbs =pos_token_freq[pos_token_freq.pos_tag == "VERB"]
verbs[:10]
```

Named entity recognition tagging

```python
ner = pd.DataFrame(columns= ["token", "ner_tag"])
for word in spacy_doc.ents:
    if pd.isna(word.label_) is False:
        ner = pd.concat([ner, pd.DataFrame.from_records([{"token": word.text, "ner_tag": word.label_ }])],ignore_index= True)
ner.head()
```

```python
#token frequency
ner_token_freq= ner.groupby(["token","ner_tag"]).size().reset_index(name="count").sort_values(by = "count", ascending = False)
ner_token_freq.head(10)
```

```python
#most common person
ner_common_person = ner_token_freq[ner_token_freq.ner_tag == "PERSON"]
ner_common_person.head(10)
```

```python
#most common places
ner_common_places = ner_token_freq[ner_token_freq.ner_tag == "GPE"]
ner_common_places.head(10)
```
