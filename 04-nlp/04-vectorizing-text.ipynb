1. Bag of Words

Bag of Words (BoW) converts text into numerical vectors by counting how many times each unique word appears in a document, completely ignoring grammar and word order.

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
countvec = CountVectorizer()
```

```python
data = pd.read_csv("04-bagofwords.txt")
```

```python
countvec_fit = countvec.fit_transform(data)
```

```python
bag_of_words = pd.DataFrame(countvec_fit.toarray(), columns = countvec.get_feature_names_out())
print(bag_of_words)
```

2. Term Frequency Inverse Document Frequency

(TF-IDF) scores each word by how often it appears in a specific document relative to how rarely it appears across the entire corpus. Keeps more of the context behind every word and measures it's importance. Documents means a single line.

Term Frequency (TF): Measures how frequently a term t appears in a single document d.

Inverse Document Frequency (IDF): Measures how rare that term is across all documents D. If a word appears in every document, log_e(1) = 0, pushing its score down.

TF-IDF Score: Multiplying the two rewards words that appear often in one document but rarely elsewhere. The higher the score the less common the word and the lower the score the common the wrod

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidfvec = TfidfVectorizer()
```

```python
tfidfvec_fit = tfidfvec.fit_transform(data)
```

```python
tfidf = pd.DataFrame(tfidfvec_fit.toarray(), columns = tfidfvec.get_feature_names_out())
print(tfidf)
```
